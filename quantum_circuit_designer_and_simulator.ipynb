```python
import ipywidgets as widgets
from IPython.display import display, clear_output

circuit_names = [
    "Identity (I)",
    "Pauli-X",
    "Pauli-Y",
    "Pauli-Z",
    "Hadamard (H)",
    "Phase-S",
    "Phase-T",
    "CNOT (CX)",
    "CZ",
    "SWAP",
    "Toffoli (CCX)",
    "Bell State",
    "GHZ State"
]

dropdown = widgets.Dropdown(
    options=circuit_names,
    description="Circuit:",
    style={'description_width': 'initial'}
)

button = widgets.Button(
    description="Run Circuit",
    button_style='primary'
)

output = widgets.Output()

def run_circuit(b):

    with output:
        clear_output()

        name = dropdown.value

        if name == "Identity (I)":
            qc = QuantumCircuit(1)
            qc.id(0)

        elif name == "Pauli-X":
            qc = QuantumCircuit(1)
            qc.x(0)

        elif name == "Pauli-Y":
            qc = QuantumCircuit(1)
            qc.y(0)

        elif name == "Pauli-Z":
            qc = QuantumCircuit(1)
            qc.z(0)

        elif name == "Hadamard (H)":
            qc = QuantumCircuit(1)
            qc.h(0)

        elif name == "Phase-S":
            qc = QuantumCircuit(1)
            qc.s(0)

        elif name == "Phase-T":
            qc = QuantumCircuit(1)
            qc.t(0)

        elif name == "CNOT (CX)":
            qc = QuantumCircuit(2)
            qc.x(0)
            qc.cx(0, 1)

        elif name == "CZ":
            qc = QuantumCircuit(2)
            qc.h(0)
            qc.h(1)
            qc.cz(0, 1)

        elif name == "SWAP":
            qc = QuantumCircuit(2)
            qc.x(0)
            qc.swap(0, 1)

        elif name == "Toffoli (CCX)":
            qc = QuantumCircuit(3)
            qc.x(0)
            qc.x(1)
            qc.ccx(0, 1, 2)

        elif name == "Bell State":
            qc = QuantumCircuit(2)
            qc.h(0)
            qc.cx(0, 1)

        elif name == "GHZ State":
            qc = QuantumCircuit(3)
            qc.h(0)
            qc.cx(0, 1)
            qc.cx(1, 2)

        print("Selected Circuit:")
        print(name)

        print("\nCircuit Diagram:")
        print(qc)


        state = Statevector.from_instruction(qc)

        print("\nStatevector:")
        print(state)


        print("\nState Visualization:")

        display(plot_bloch_multivector(state))


        measured_qc = qc.copy()

        measured_qc.measure_all()

        simulator = AerSimulator()

        result = simulator.run(
            measured_qc,
            shots=1024
        ).result()

        counts = result.get_counts()

        print("\nMeasurement Results:")
        print(counts)


        print("\nMeasurement Histogram:")

        plot_histogram(counts)
        plt.show()

button.on_click(run_circuit)

display(dropdown)
display(button)
display(output)
```
